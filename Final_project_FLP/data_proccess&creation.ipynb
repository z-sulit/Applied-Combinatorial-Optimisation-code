# Data preproccess and Synthetic Data (nodes)

```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist
from haversine import haversine
```

```python
shops_df = pd.read_csv("davao_coffee_matcha_shops.csv")
np.random.seed(42)
# (simulates a few high-volume cafes and many average-volume cafes)
base_demand = np.random.lognormal(mean=4.5, sigma=0.45, size=len(shops_df))
```

```python
# Approximate coordinates for Davao City's downtown core (Poblacion)
core_lat, core_lon = 7.075, 125.605
dist_from_center = np.sqrt((shops_df["latitude"] - core_lat)**2 + (shops_df["longitude"] - core_lon)**2)
proximity_factor = np.clip(1.3 - (dist_from_center * 3), 0.7, 1.3)
# Calculate final demand weight and round to nearest integer
shops_df["demand_weight"] = np.round(base_demand * proximity_factor).astype(int)
```

```python
shops_df["ID"] = range(len(shops_df))
shops_df["LE_ENGLISH"] = shops_df["place_name"]
shops_df["POPULATION"] = shops_df["demand_weight"]
shops_df["Y"] = shops_df["latitude"]
shops_df["X"] = shops_df["longitude"]
#--------%%%%%%
nodes_cols = ["ID", "place_name", "latitude", "longitude", "demand_weight", "LE_ENGLISH", "X", "Y", "POPULATION"]
nodes_df = shops_df[nodes_cols]
nodes_df.to_csv("davao_coffee_matcha_nodes.csv", index=False)
print("davao_coffee_matcha_nodes.csv done!")
```
